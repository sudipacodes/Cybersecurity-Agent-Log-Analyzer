```python
import pandas as pd
df = pd.read_csv("logs.csv")
```

```python
print(" Logs Loaded Successfully")
print(df.head())
```

```python
# Compress logs by grouping similar events
compressed_logs = df.groupby(["user","action","status"]).size().reset_index(name="count")

print(" Original Log Size:", len(df))
print(" Compressed Log Size:", len(compressed_logs))

print("\nCompressed Logs Preview:")
print(compressed_logs.head())
```

```python
# Convert categorical data to numbers
df_encoded = df.copy()

df_encoded["user"] = df_encoded["user"].astype("category").cat.codes
df_encoded["ip"] = df_encoded["ip"].astype("category").cat.codes
df_encoded["action"] = df_encoded["action"].astype("category").cat.codes
df_encoded["status"] = df_encoded["status"].astype("category").cat.codes

print(" Logs Encoded for AI")
df_encoded.head()
```

```python
from sklearn.ensemble import IsolationForest

# Select features
features = df_encoded[["time","user","ip","action","status"]]

# Train anomaly detection model
model = IsolationForest(contamination=0.15, random_state=42)
df_encoded["anomaly"] = model.fit_predict(features)

print(" Threat Detection Completed")
df_encoded.head()
```

```python
# Find suspicious activities
alerts = df[df_encoded["anomaly"] == -1]

print(" ALERTS DETECTED:")
print(alerts)
```

```python
import matplotlib.pyplot as plt

# Count anomalies
counts = df_encoded["anomaly"].value_counts()

plt.bar(counts.index.astype(str), counts.values)
plt.title("Threat Detection Results")
plt.xlabel("Anomaly (-1 = Threat, 1 = Normal)")
plt.ylabel("Number of Events")
plt.show()
```
